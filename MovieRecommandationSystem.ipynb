```python
import pandas as pd
from pathlib import Path

data_path = Path(path)
df_movie = pd.read_csv(data_path / "tmdb_5000_movies.csv")
df_credits = pd.read_csv(data_path / "tmdb_5000_credits.csv")
df_movie.info()
```

```python
movies = df_movie.merge(df_credits, on="title")
movies.info()
```

```python
# important cols
features = ["id", "genres", "keywords", "title", "overview", "cast", "crew"]
movies = movies[features]
movies.info()
```

```python
# convert data to list of str

def convert_cast(obj):
  count = 0
  obj_list = []
  for i in eval(obj):
    if count != 3:
      obj_list.append(i["name"])
      count += 1
    else:
      break
  return obj_list

def convert_crew(obj):
  obj_list = []
  for i in eval(obj):
    if i["job"] == "Director":
      obj_list.append(i["name"])

  return obj_list

movies["genres"] = movies["genres"].apply(lambda x: [ i["name"]  for i in eval(x)])
movies["keywords"] = movies["keywords"].apply(lambda x: [ i["name"]  for i in eval(x)])
movies["cast"] = movies["cast"].apply(lambda x: convert_cast(x))
movies["crew"] = movies["crew"].apply(lambda x: convert_crew(x))
movies["overview"] = movies["overview"].apply(lambda x: x.split())

```

```python
# remove blank spaces
def remove_blankspace(list_str):
  return [i.replace(" ", "") for i in list_str]


movies["cast"] = movies["cast"].apply(remove_blankspace)
movies["crew"] = movies["crew"].apply(remove_blankspace)
movies["genres"] = movies["genres"].apply(remove_blankspace)
movies["keywords"] = movies["keywords"].apply(remove_blankspace)
```

```python
print(movies.iloc[0].cast)
print(movies.iloc[0].crew)
print(movies.iloc[0].genres)
print(movies.iloc[0].keywords)
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']

dataset = movies[["id", "title", "tags"]].copy()
dataset.head()
```

```python
dataset["tags"] = dataset["tags"].apply(lambda x: " ".join(x))
dataset["tags"] = dataset["tags"].apply(lambda x: x.lower())
```

```python
dataset["tags"][0]
```

```python
# vectorization

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english"
)

vectors = tfidf.fit_transform(dataset["tags"]).toarray()
```

```python
vectors
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python


def recommand_movie(movie_title):
  movie_index = dataset[dataset["title"] == movie_title].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
  recommanded_movies = []
  for movie in movies_list:
    recommanded_movies.append(dataset.iloc[movie[0]].title)
  return recommanded_movies

```

```python
recommand_movie("Batman")
```
